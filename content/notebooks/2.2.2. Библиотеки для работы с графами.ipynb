Граф — это набор **вершин** и **рёбер** между ними. В Python с графами удобно работать через библиотеку **NetworkX**.

Скачайте эту тетрадку и выполняйте ячейки по порядку: сначала запустите блок с подготовкой, затем впишите ответы в ячейки заданий и проверьте себя.

> Если NetworkX не установлен, выполните в терминале: `pip install networkx`

```python
import networkx as nx

# Пример: маленький неориентированный граф
demo = nx.Graph()
demo.add_edges_from([('X', 'Y'), ('Y', 'Z')])

print('Вершины:', list(demo.nodes))
print('Рёбра:  ', list(demo.edges))
print('Степени:', dict(demo.degree()))
```
