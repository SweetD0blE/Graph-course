Граф — это набор **вершин** и **рёбер**. Ниже уже построен небольшой граф **G** на библиотеке **NetworkX**. Изучите его, выполните ячейку кода, а затем ответьте на вопросы — короткий ответ вводится в поле под вопросом.

> Если NetworkX не установлен, выполните: `pip install networkx`

```python
import networkx as nx

# Граф G: 4 вершины и 4 ребра
G = nx.Graph()
G.add_nodes_from(['A', 'B', 'C', 'D'])
G.add_edges_from([('A', 'B'), ('B', 'C'), ('C', 'D'), ('A', 'C')])

print('Вершины:', list(G.nodes))
print('Рёбра:  ', list(G.edges))
print('Степени:', dict(G.degree()))
```
